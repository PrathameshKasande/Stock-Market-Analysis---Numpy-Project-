Stock Market Analysis Project Using NumPy

Objectives 

- Analyze stock closing prices using NumPy.

- Perform basic and statistical analysis, filtering, broadcasting, and memory optimization,Reshapping,Flattening.




```python
import numpy as np
# stock Closing prices  for 10 days of single company 
close_price = np.array([150, 152, 149, 153, 155, 157, 156, 160, 162,161])
```

Basic Analysis 


```python
print("Shape of Stock Closing price:",close_price.shape) # shape(row X columns )
print("Size of the Stock Close Prices :",close_price.size) #Total Elements in the array
print("Dimension of stock close price :",close_price.ndim)
print("datatype of close price of stock:",close_price.dtype)
print("Minimum Stock closing Price in !0 days :","Rs.",np.min(close_price))
print("Maximum Stock closing price in 10 days : ","Rs.",np.max(close_price))
print("Sum of closing price of 10 days :","Rs.",np.sum(close_price))
print("For 2nd day closing price of stock: ","Rs.",close_price[00]) # slicing 

```

 Memory Before Optimzation 

```python
print("Total Memory (bytes):",close_price.nbytes,"Bytes ") # nbytes gives itemsize × size 
print("Total Memory  for each close price (bytes):",close_price.itemsize,"Bytes ") # gives bytes per element 
```

Memory Optimization and Changing Data type  


```python
optimize=close_price.astype(np.float16) #changes int datatype into float and optimize the bytes 
print("Original data Close_price Array:",close_price.dtype)
print("Optimized Datatype of close price Array:",optimize.dtype)
print("Optimized Memory bytes :",optimize.nbytes,"Bytes")
```

Statistical Analysis for Stock Price 

```python
print("Average of 10 days Stocks of 10 days  Closing Price :","RS.",np.mean(close_price))
print("Median stock of 10 days closing price: ","RS.",np.median(close_price))
print("Standard deviation for 10 days stock closing price:","RS.",np.std(close_price))
print("Variance having  stock of 10 days closing price:","RS.",np.var(close_price)) 

```

Daily Returns After  Closing The Stock 

```python
daily_returns = np.diff(close_price) / close_price[:-1] * 100 #finds difference between each consecutive day
print("Everday Returns after closing the  stocks :\n ",daily_returns)
print("Daily Returns (%):", np.round(daily_returns, 2)) # round gives rounded values 
```

Filtering the data 

```python
high_price=close_price[close_price>156]
print("stock peak values in 10 days : ",high_price,"Rs")

low_price=close_price[close_price<150]
print("Lowest stock closing price in 10 days :",low_price,"Rs")
```

Conditional Filtering (If-Else Condition) for Alerts

```python
today_price = close_price[-1] # indexing negative 
if today_price > np.mean(close_price):
    print(" Increase The stock Price of today ")
else:
    print("Possible the Downfall and Decreases the stock closing Price ")
```

Reshape the Stock Close Price Array 

```python
Reshaped_arr=close_price.reshape(2, 5) # reshape the original array of 1D to multiDimensional 
print("reshaped array of stock close prices :\n",Reshaped_arr)
```

Flatten the Array 

```python
Flattened_arr=Reshaped_arr.flatten( ) # convert reshaped 2d array into 1 dimensional array 
print("Flattened array of Stock Closing Prices :\n ",Flattened_arr)
```

Broadcasting 
Grow the Stock over 10 days 

```python
# stock prices increases by rs.10 (bonus adjustments)

New_prices=close_price+ (10 )
print("All Stock Prices After 10 days :","Rs.",np.round(New_prices,0))
```
